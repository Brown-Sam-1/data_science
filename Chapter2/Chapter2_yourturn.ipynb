## Question

You want to send a document to your bank and it is very important it gets there the next day. You therefore send two copies of the document using Hermes (delivery service $A$) and the Royal Mail (delivery service $B$).  The probabilities of those companies delivering on time are:

- 0.90 for service A
- 0.80 for service B
- 0.75 for both services being on time.

Does your strategy improve the chances the document will be delivered on time?

### Answer

*Your answer here.*

Letting:

* Service A: $P(A)=0.9$
* Service B: $P(B)=0.8$
* Both services: $P(A\cap B)=0.75$

Using addition rule: $$P(A\cup B)=P(A)+P(B)-P(A\cap B)$$ $$=0.9+0.8-0.75=0.95$$
So it does improve the chances.

## <font color=#30a64d>Question</font>

You flip a coin 5 times. What is the probability that heads will not occur 3 times?    

### Answer

*Your answer here.*

Doing 5 choose 3 tells us how many ways there are to get 3 heads/tails:
$$\begin{pmatrix} 5\\3 \end{pmatrix} =10$$
Now the chance to get 5 things you want in a row is:
$$(\frac{1}{2})^5=\frac{1}{32}$$
So the probability of getting 3 heads is:
$$P(3 \text{ heads})=\frac{10}{32}=\frac{5}{16}$$
So the probability of not getting 3 heads is the complement:
$$P(\text{not } 3 \text{ heads})=1-\frac{5}{16}=\frac{11}{16}=68.75\%$$

# Bayes Theorem

## <font color=#30a64d>Question</font>

Suppose you think you have a rare disease where 1 in 10,000 people are affected. You go to your doctor, and she performs a test. The test is of a high quality, and is correct 99% of the time.

Unfortunately your results come back positive for the disease. What is your chance of having the disease? Is it (roughly): <br><br>

a) 99% <br>
b) 90% <br>
c) 10% or <br>
d) 1%? <br> <br>

Repeat the same as above, but change 1 in 10,000 to 1 in 100. What would the chances of having the disease be now?

### Answer

*Your answer here.*

To visualise the problem, below is a flowchart:

```python
print("To visualise the problem, below is a flowchart:")
import graphviz

dot = graphviz.Digraph(comment='1 in 10000 disease')
dot.node('A', '10000 people', shape='square')
dot.node('B', '1 with disease', shape='square')
dot.node('C', '9999 without disease', shape='square')
dot.node('D', 'true positive', shape='square')
dot.node('E', 'false negative', shape='square')
dot.node('F', 'false positive', shape='square')
dot.node('G', 'true negative', shape='square')

dot.edge('A', 'B', label='1/10000')
dot.edge('A', 'C', label='9999/10000')
dot.edge('B', 'D', label='0.99')
dot.edge('B', 'E', label='0.01')
dot.edge('C', 'F', label='0.01')
dot.edge('C', 'G', label='0.99')

dot
```

The Bayesian formula to get the probability that someone has the disease given that the test came out as positive is as follows:
$$P(D|+ve)=\frac{P(+ve|D)P(D)}{P(+ve)}$$
Where:
$$P(+ve|D)=0.99$$
$$P(D)=0.0001$$
$$P(+ve)=P(+ve|D)P(D)+P(+ve|\text{no }D)P(\text{no }D)$$
Therefore the probability is:
$$P(D|+ve)=\frac{(0.99)(0.0001)}{(0.99)(0.0001)+(0.01)(0.9999)}=\frac{1}{102}\approx1\%$$

When changing the disease chance to:
$$P(D)=0.01$$
The probability then becomes:
$$P(D|+ve)=\frac{(0.99)(0.01)}{(0.99)(0.01)+(0.01)(0.99)}=\frac{1}{2}=50\%$$
